## importing all frameworks

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

#### creating folders and checking if data exists and importing data

```python
cur_dir = Path.cwd()

data_dir = Path(cur_dir/"data"/"raw")
data_dir.mkdir(exist_ok=True)
file_path = data_dir / "Titanic-Dataset.csv"
try:
    df = pd.read_csv(file_path)
    print(f"✅ Loaded {len(df)} rows and {len(df.columns)} columns")
    print(f"📄 File: {file_path.name}")
except FileNotFoundError:
    print("❌ File not found")
    print(f"📁 Expected folder: {data_dir}")
    print(f"📄 Expected file:   {file_path.name}")
    print("👉 Put the CSV in that folder and run this cell again.")
```

```python
df["Age"] = df["Age"].fillna(
    df.groupby(["Pclass", "Sex"])["Age"].transform("median")
)
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["FamilySize"].min(), df["FamilySize"].max()
df[["SibSp", "Parch", "FamilySize"]]
```

```python
df.groupby("Sex")["Survived"].mean()
df.groupby("Pclass")["Survived"].mean()
df.groupby(["Pclass","Sex"])["Survived"].mean()
```
